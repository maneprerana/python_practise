## Fibonacci 

```python
def fibonacci(n):
    a, b = 0, 1
    sequence = []

    for _ in range(n):
        sequence.append(a)
        a, b = b, a + b

    return sequence

print(fibonacci(10))
```

```python
num = int (input("Enter a number : "))
original = num
reverse = 0


while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse :
    print("Palindrome Number")
else:
     print("Not a Palindrome Number")
```
